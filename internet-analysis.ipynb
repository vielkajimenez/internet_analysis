```python
#import and inspect data

#entity, the name of the country, region, or income bracket
#code, the three-letter country code
# year, the year
#internet_users_per_100, the number of internet users for every 100 people in the entity's population


internet = pd.read_csv('internet.csv')
internet.head()
```

```python
internet.rename(columns={'internet_users_per_100':'percent_online'}, inplace=True)
internet.info() 
```

```python
#Years to Reach Mainstream Use
#Calculates how long it takes for internet use to reach the majority of people in an entity using the percent_online column. The function will return a string indicating whether the entity has no internet use, few people online, some people online, or most people online.

def internet_use_by_entity(row):
    if row['percent_online'] == 0:
        return 'none'
    elif row['percent_online'] > 0 and row['percent_online'] < 25:
        return 'few'
    elif row['percent_online'] >= 25 and row['percent_online'] < 50:
        return 'some'
    else:
        return 'most'
    
#Create a new column named amount_online on internet by applying the function internet_use_by_entity to each row
    
internet['amount_online'] = internet.apply(internet_use_by_entity, axis=1)
internet.head()
```
